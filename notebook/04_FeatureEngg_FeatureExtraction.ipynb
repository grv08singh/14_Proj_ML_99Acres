```python
# handling nearby_loc feature
import ast

df['nearby_1'] = df['nearby_loc'].apply(
    lambda x: ast.literal_eval(x) if isinstance(x, str) else x
)

df['landmark_count'] = df['nearby_1'].apply(lambda x: len(x))
```

```python
from collections import Counter

stopwords = {'the', 'a', 'an', 'and', 'of', 'in', 'at', 'to', 'is', 'for', 'on'}

all_words = []
for landmark_list in df['nearby_1']:
    for landmark in landmark_list:
        # Lowercase, remove special characters, split into words
        words = re.sub(r'[^a-z0-9\s]', '', landmark.lower()).split()
        filtered = [w for w in words if w not in stopwords and len(w) > 2]
        all_words.extend(filtered)

word_freq = Counter(all_words)
```

```python
top_n = 100
nearby_freq = pd.DataFrame(word_freq.most_common(top_n), columns=['word', 'frequency'])
```

```python
not_landmark = ('gurgaon','sector','gandhi','indira','dwarka','gurugram','sohna',
                'igi','delhi','manesar','imt','5556','sec')

general_landmarks = nearby_freq[~nearby_freq['word'].isin(not_landmark)].iloc[:,0].tolist()
```

```python
landmark_dict = {
    'has_hospital_or_clinic': ['hospital','nursing','clinic','health care'],
    'has_police_stn': ['police'],
    'has_fire_stn': ['fire'],
    'has_mall': ['mall'],
    'has_market': ['pharmacy','market'],
    'has_park': ['park'],
    'has_bank_or_atm': ['atm','bank'],
    'has_institute': ['institute','school','vidyalaya','college','university','academy'],
    'has_airport': ['airport'],
    'has_rly_stn': ['railway station','rly station','railway stn','rly stn'],
    'has_metro_stn': ['metro'],
    'has_expy': ['expressway','expy'],
    'has_petrol_pump': ['petrol','cng','oil','petroleum'],
    'has_hotel': ['hotel','inn','radisson','lemontree'],
    'has_temple_or_church': ['temple','mandir','church'],
    'has_famous_chowk': ['chowk'],
    'has_club': ['club'],
    'has_stadium': ['stadium'],
    'has_restro_cafe': ['restaurant','cafe','food court','dhaba','burger','pizza','dominos','mcdonald','coffee']
}
```

```python
for key, values in landmark_dict.items():
    df[key] = 0
    for index, df_nearby in df.iterrows():
        landmarks = df_nearby['nearby_loc'].replace("[",'').replace("]",'').replace("'",'').split(",")
        landmarks = [landmark.strip() for landmark in landmarks if len(landmark.strip())>2]
        for landmark in landmarks:
            for val in values:
                if val in landmark:
                    df.loc[index,key] = 1
                    break
```

```python
for row in df['furnishDetails']:
    row = str(row).lower().replace('[','').replace(']','').replace("'","").replace('no','0').strip().split(",")
    for item in row:
        if item != 'nan':
            name_obj = re.findall(r"([a-zA-Z/s]+)", item)
            if name_obj is not None:
                name = " ".join(name_obj)
                
                count_obj = re.search(r"(\d+)", item)
                if count_obj is not None:
                    count = count_obj.group()
                    df[name]=count
                else:
                    df[name]=0

df.drop(columns=['nearby_1','nearby_loc','furnishDetails'], inplace=True)
```
