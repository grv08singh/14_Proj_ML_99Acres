```python
# handling nearby_loc feature
import ast

df['nearby_1'] = df['nearby_loc'].apply(
    lambda x: ast.literal_eval(x) if isinstance(x, str) else x
)

df['landmark_count'] = df['nearby_1'].apply(lambda x: len(x))
```

```python
from collections import Counter

stopwords = {'the', 'a', 'an', 'and', 'of', 'in', 'at', 'to', 'is', 'for', 'on'}

all_words = []
for landmark_list in df['nearby_1']:
    for landmark in landmark_list:
        # Lowercase, remove special characters, split into words
        words = re.sub(r'[^a-z0-9\s]', '', landmark.lower()).split()
        # Filter stopwords and short tokens
        filtered = [w for w in words if w not in stopwords and len(w) > 2]
        all_words.extend(filtered)

word_freq = Counter(all_words)
```

```python
top_n = 100
nearby_freq = pd.DataFrame(word_freq.most_common(top_n), columns=['word', 'frequency'])
```

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('all-MiniLM-L6-v2')

def embed_landmarks(landmark_list):
    embeddings = model.encode(landmark_list)       # shape: (n_landmarks, 384)
    return np.mean(embeddings, axis=0)             # mean pooling → (384,)

landmark_vecs = np.vstack(df['landmarks'].apply(embed_landmarks))
```

```python
for row in df['furnishDetails']:
    row = str(row).replace('[','').replace(']','').replace("'","").replace('no','0').strip().split(",")
    for item in row:
        if item != 'nan':
            name_obj = re.findall(r"([a-zA-Z/s]+)", item)
            if name_obj is not None:
                name = " ".join(name_obj)
                
            count_obj = re.search(r"(\d+)", item)
            if count_obj is not None:
                count = count_obj.group()
                df[name]=count

df.drop(columns=['furnishDetails'], inplace=True)
```
